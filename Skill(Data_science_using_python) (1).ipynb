# Experiment 2:Data Preprocessing - Cleaning dataset

## 1. Handling Missing Values

```python
import pandas as pd

mergeddataset = mergeddataset.dropna()
mergeddataset
```

## 2. Removing Duplicate

```python
mergeddataset = mergeddataset.drop_duplicates()
```

## 3. Handling Outliers

```python
import pandas as pd


q1 = mergeddataset["alcohol"].quantile(0.25)
q3 = mergeddataset["alcohol"].quantile(0.75)
iqr = q3 - q1
df_clean= mergeddataset[(mergeddataset["alcohol"] > q1 - 1.5 * iqr) & (mergeddataset["alcohol"] < q3 + 1.5 * iqr)]
df_clean
```

# Experiment 7 & 8 : Filter Methods Of Feature Selection

## Variance Threshold

```python
from sklearn.feature_selection import VarianceThreshold

feature_scores = {}

x = df.drop(columns=['type','label_type','quality_encoded','target_type','freq_type','MinMaxAlcohol','alcohol_standardized'])  # Features
y = df['type']  # Target variable

selector = VarianceThreshold()
x_variance_filtered = selector.fit_transform(x)

variance_scores = pd.Series(selector.variances_, index=x.columns)
feature_scores["Variance"] = variance_scores

feature_scores["Variance"]
```

## Pearson's Coefficient

```python
corr_matrix = x.corr().abs()
pearson_scores = corr_matrix.mean(axis=1)  # Mean correlation for each feature
feature_scores["Correlation"] = pearson_scores
feature_scores["Correlation"]
```

## Chi-Square

```python
from sklearn.feature_selection import SelectKBest, chi2

X_cat = x.select_dtypes(include=['float64', 'object'])
X_cat = pd.get_dummies(X_cat)
if X_cat.shape[1] > 0:
    chi2_selector = SelectKBest(chi2, k=min(5, X_cat.shape[1]))
    chi2_selector.fit(X_cat, y)
    chi2_scores = pd.Series(chi2_selector.scores_, index=X_cat.columns)
    feature_scores["Chi-Square"] = chi2_scores

feature_scores["Chi-Square"]
```

## Information Gain

```python
from sklearn.feature_selection import mutual_info_classif

mi_scores = mutual_info_classif(x, y)
mi_scores_series = pd.Series(mi_scores, index=x.columns)
feature_scores["Mutual Information"] = mi_scores_series

feature_scores["Mutual Information"]
```

# Experiment 9 & 10 : Wrapper Methods Of Feature Selection

## Forward Feature Selection

```python
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import LogisticRegression

# Define model
model = LogisticRegression(max_iter=1000)

# Apply Forward Selection
sfs = SequentialFeatureSelector(
    estimator=model,
    direction='forward',
    n_features_to_select=5,
    scoring='accuracy',
    cv=2,
    n_jobs=-1

)
sfs.fit(x, y)

# Get selected features
selected_features = x.columns[sfs.get_support()]
feature_scores = pd.DataFrame({
    "Feature": x.columns,
    "Selected": sfs.get_support()
})

print(feature_scores)
```

## Backward Feature Elimination with Scores

```python
sbs = SequentialFeatureSelector(model, n_features_to_select=5, direction='backward',n_jobs=-1)
sbs.fit(x, y)

# Get selected features
selected_features = x.columns[sbs.get_support()]
feature_scores = pd.DataFrame({
    "Feature": x.columns,
    "Selected": sbs.get_support()
})

print(feature_scores)
```

# Experiment 11 : Embeded Methods Of Feature Selection

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(penalty='l1', solver='liblinear', max_iter=1000)
model.fit(x, y)
coef_abs = np.abs(model.coef_[0])

top_indices = np.argsort(coef_abs)[-5:][::-1]

top_features = x.columns[top_indices]

print("Top 5 Features :")
print(top_features)
```

# 12 :Model Implementation & Ensemble Learning

## Spliting & Model Implementation

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x , y , test_size=0.2 , random_state=42)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)
print(f"Logistic Regression Accuracy: {accuracy:.4f}")
```

## Ensemble learning

### Bagging

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier

base_model = DecisionTreeClassifier()

bagging_model = BaggingClassifier(
    estimator=base_model,
    n_estimators=10,
    max_samples=0.8,
    random_state=42,
    n_jobs=-1
)

bagging_model.fit(X_train, y_train)

score = bagging_model.score(X_test, y_test)

print(f"BaggingClassifier (Decision Tree Model) Score : {score:.3f} ")
```

### Boosting

```python
from sklearn.ensemble import AdaBoostClassifier
base_model = DecisionTreeClassifier()

boosting_model = AdaBoostClassifier(
    estimator=base_model,
    n_estimators=10,
    random_state=42
)

boosting_model.fit(X_train, y_train)
y_pred = boosting_model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy of AdaBoost Model: {accuracy:.4f}")
```
